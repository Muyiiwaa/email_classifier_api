```python
data_zero = data[data['is_churned'] == 0].iloc[0:2071]
data_one = data[data['is_churned'] == 1]

data = pd.concat([data_zero, data_one])
data.reset_index(drop=True, inplace=True)




X = data.drop(columns=["is_churned"])
y = data['is_churned']

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder


categorical_cols = data.select_dtypes(include="object").columns
for col in categorical_cols:
    encoder = LabelEncoder()
    data[col] = encoder.fit_transform(data[col])
    

X_train, X_test, y_train, y_test = train_test_split(X,y, random_state=23,test_size=0.5,
                                                    stratify=y)
```

```python
scaler = StandardScaler()
column_names = X_train.columns
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)

X_train = pd.DataFrame(data=X_train, columns = column_names)
X_test = pd.DataFrame(data=X_test, columns = column_names)
```

```python
model = RandomForestClassifier(random_state=23, n_estimators=5)
from sklearn.metrics import f1_score


model.fit(X_train, y_train)
train_preds = model.predict(X_train)
test_preds = model.predict(X_test)

print(f1_score(y_train, train_preds))
print(f1_score(y_test, test_preds))
```
